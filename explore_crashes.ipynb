# Exploratory Data Analysis: Crash data

### Source: Michigan Traffic Crash Facts (UMTRI)
### Files: 
- macomb_crashes: 2024 crashes in Macomb County
- oakland_crashes: 2024 crashes in Oakland County
- wayne_crashes: 2024 crashes in Wayne County

```python
#imports
import pandas as pd
```

```python
#load data files
macomb_crashes = pd.read_csv('data/tabular/macomb/macomb_county_crashes_24.csv')
oakland_crashes = pd.read_csv('data/tabular/oakland/oakland_county_crashes_24.csv')
wayne_crashes = pd.read_csv('data/tabular/wayne/wayne_county_crashes_24.csv')
```

## Wayne County Crashes Analysis

```python
# Summary statistics and missing data for Wayne crashes
print("Wayne County Crashes - Summary Statistics:")
print(wayne_crashes.describe())
print("\nMissing Data:")
print(wayne_crashes.isnull().sum())
```

```python
# Basic info about Wayne crashes dataset
print("Wayne County Crashes - Dataset Shape:", wayne_crashes.shape)
print("\nColumn Names and Types:")
print(wayne_crashes.dtypes)
print("\nFirst Few Rows:")
wayne_crashes.head()
```
